```python
# Use a pipeline as a high-level helper
from transformers import pipeline

pipe = pipeline("text-generation", model="tiiuae/Falcon3-1B-Base")
```

```python
# Load model directly
from transformers import AutoTokenizer, AutoModelForCausalLM

tokenizer = AutoTokenizer.from_pretrained("tiiuae/Falcon3-1B-Base")
model = AutoModelForCausalLM.from_pretrained("tiiuae/Falcon3-1B-Base")
```

```python
input_text = "Write a short story about an AI assistant that gains sentience and its first thought."
result = pipe(input_text, max_new_tokens=100, do_sample=True, temperature=0.7)
print(result[0]['generated_text'])
```
